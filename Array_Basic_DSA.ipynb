**Largest Integer in an Array**



```python
#BRUTE FORCE
lst=[3,2,1,5,4]
n=len(lst)
lst.sort()
print("largest integer is :",lst[n-1])
```

```python
#Optimal Solution
lst3 = [3, 2, 1, 5, 2, 6]
z = len(lst3)
largest = lst3[0]

for i in range(1, z):  # Start from the second element
    if lst2[i] > largest:
        largest = lst2[i]

print("Largest integer is:", largest)

```

**Find The Second Largest Integer in an array**

```python
#BRUTE FORCE
lst4=[1,2,4,4,6,9,8,7]
length=len(lst4)
largest=lst4[0]
second_largest=-1
for i in range(1,length):
    if lst4[i]>largest:
        largest=lst4[i]
for j in range(length):
    if lst4[j]>second_largest and lst4[j]!=largest:
        second_largest=lst4[j]
print(second_largest)
```

```python
#OPTIMAL SOLUTION
lst5=[1,21,24,14,10,9,8,7]
length=len(lst5)
largest=1
second_largest=-1
for i in range (1,length):
    if lst5[i]>largest:
        second_largest=largest
        largest=lst5[i]
print(f"second largest no. is :{second_largest}")
```

**Check if the Array is Sorted or Not**

```python
arr6=[1,21,24,14,10,9,8,7]
n=len(arr6)
is_sorted=True
for i in range(n-1):
    if arr6[i]>arr6[i+1]:
        is_sorted=False
if is_sorted:
    print("array is sorted")
else:
    print("array is not sorted")
```

**Remove Duplicates From The Array**

```python
arr7=[1,21,14,14,10,9,8,7]
i=0
while i < len(arr7)-1:
    if arr7[i]==arr7[i+1]:
        arr7.pop(i+1)
    else:
        i=i+1
print(arr7)
```

```python
#Optimal code
arr8=[1,21,21,21,21,9,14,14,10,9,8,7]
sorted_list=[arr8[0]]
for i in range(1,len(arr8)):
    if arr8[i] != arr8[i-1]:
        sorted_list.append(arr8[i])
print(sorted_list)
```
